## Specify the name of the podcast and the sources

Sources can be URLs to web articles or YouTube videos, or local files like PDF, TXT, etc.
We don't support all file types yet, like .docx for example, but we're working on it!

```python
from neuralnoise.models import StudioConfig

podcast_name = "The NeuralNoise Podcast"
sources = [
    "https://learnhub.top/celebrating-21-years-of-notepad-the-legendary-journey-of-our-favorite-text-editor/",
    "https://www.apple.com/newsroom/2024/10/new-macbook-pro-features-m4-family-of-chips-and-apple-intelligence/",
    "https://openai.com/index/introducing-chatgpt-search/",
]

config = StudioConfig(
    **{
        "show": {
            "name": podcast_name,
            "about": "A podcast about the future of AI and the impact it will have on our lives. We discuss interesting topics in AI and technology.",
            "language": "English",
            "min_segments": 4,
            "max_segments": 6,
        },
        "speakers": {
            "speaker1": {
                "name": "Zach",
                "about": "Zach is a software engineer and the founder of NeuralNoise.",
                "settings": {
                    "provider": "openai",
                    "voice_model": "tts-1",
                    "voice_id": "alloy",
                },
            },
            "speaker2": {
                "name": "Emily",
                "about": "Emily is the co-host of NeuralNoise and a product designer.",
                "settings": {
                    "provider": "openai",
                    "voice_model": "tts-1",
                    "voice_id": "nova",
                },
            },
        },
    }
)
```

```python
from neuralnoise import aextract_content

# Extract content from the sources
content = await aextract_content(sources)

# Show an example of the content
content[:100]
```

```python
from neuralnoise import PodcastStudio

studio = PodcastStudio(work_dir="output", config=config)

# Generate a script for the podcast episode
script = studio.generate_script(content)

print(f"Generated script with {len(script['sections'])} sections")
```

```python
# Create a podcast episode
podcast_episode = studio.generate_podcast_from_script(script)
```

```python
from IPython.display import Audio

podcast_episode.export("output.wav", format="wav")
Audio("output.wav", autoplay=True)
```
